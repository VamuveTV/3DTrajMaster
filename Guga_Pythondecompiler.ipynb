```python
# Step 1: Mount Google Drive
# This connects the Colab environment to your Google Drive.
from google.colab import drive
import os
import subprocess

print("Mounting Google Drive...")
drive.mount('/content/drive')
print("Drive mounted successfully.")

# Step 2: Install and build pycdc, a decompiler that supports more Python versions.
print("\nInstalling and building pycdc...")
# Clone the pycdc repository from GitHub.
!git clone https://github.com/zrax/pycdc.git
# Change to the pycdc directory.
%cd pycdc
# Use CMake to configure the build files.
!cmake .
# Compile the tool using make.
!make
# Change back to the content directory.
%cd /content
print("pycdc installed and built successfully.")

# --- USER CONFIGURATION ---
# Step 3: Define your input and output directories on Google Drive.
# Replace these paths with your own if they are different.
# The script will look for .pyc files in the 'input_dir' and all its subfolders.
input_dir = '/content/drive/MyDrive/PythonDecompiler/Input'
output_dir = '/content/drive/MyDrive/PythonDecompiler/Output'
# --------------------------

# Step 4: Ensure the output directory exists
# This creates the output folder if it doesn't already exist.
print(f"\nCreating output directory: {output_dir}")
os.makedirs(output_dir, exist_ok=True)
print("Output directory is ready.")

# Step 5: Walk through the input directory and decompile all .pyc files
print(f"\nStarting decompilation from {input_dir}...")
count_decompiled = 0
count_failed = 0

# os.walk() generates the file names in a directory tree
# by walking the tree either top-down or bottom-up.
for root, dirs, files in os.walk(input_dir):
    # Construct the relative path to maintain the directory structure.
    rel_path = os.path.relpath(root, input_dir)

    # Create the corresponding output subdirectory.
    output_sub_dir = os.path.join(output_dir, rel_path)
    os.makedirs(output_sub_dir, exist_ok=True)

    # Iterate through all files in the current directory.
    for file in files:
        # Check if the file is a .pyc file.
        if file.endswith('.pyc'):
            pyc_file_path = os.path.join(root, file)
            py_file_name = file.replace('.pyc', '.py')
            output_py_path = os.path.join(output_sub_dir, py_file_name)

            print(f"Decompiling {pyc_file_path} to {output_py_path}...")

            # Use subprocess to run the pycdc command.
            try:
                # pycdc outputs to stdout, so we need to redirect it to a file.
                with open(output_py_path, 'w') as f:
                    # The 'check=True' argument raises an exception if the command fails.
                    subprocess.run(['/content/pycdc/pycdc', pyc_file_path], stdout=f, check=True)
                count_decompiled += 1
                print(f"  -> SUCCESS: Decompiled to {output_py_path}")
            except subprocess.CalledProcessError as e:
                count_failed += 1
                print(f"  -> ERROR: Failed to decompile {pyc_file_path}. Error: {e}")

print("\nDecompilation process finished.")
print(f"Summary: {count_decompiled} files decompiled successfully, {count_failed} files failed.")
print(f"You can now check the output folder on your Google Drive: {output_dir}")
```
