```python
import os
import numpy as np
import nltk
import scipy.io.wavfile as wavfile
from bark import SAMPLE_RATE, preload_models
from bark.generation import generate_text_semantic
from bark.api import semantic_to_waveform
from IPython.display import Audio
from pydub import AudioSegment
from google.colab import drive

# 1. Preparação de Ambiente
nltk.download('punkt')
nltk.download('punkt_tab')
preload_models()

# Montar Drive e criar pasta
try:
    drive.mount('/content/drive')
    path_drive = "/content/drive/MyDrive/AI_Voice_Guga/Output"
    os.makedirs(path_drive, exist_ok=True)
except:
    print("⚠️ Não foi possível montar o Drive. Salvando localmente no Colab.")
    path_drive = "/content"

# 2. Roteiro
script = """
[solicitous] Hey there, TV Nostalgia crew, I'm Guga. Nice to meet you.
[clear] I'm the hunter of lost audiovisual relics... [laughs]
[enthusiastic] Yeahhh! For those who don't know me, a huge welcome to the channel.
[clear] Here we rescue those old films, those classic works that marked generations with their old-school dubbing!
""".replace("\n", " ").strip()

sentences = nltk.sent_tokenize(script)
SPEAKER = "v2/en_speaker_6"
silence = np.zeros(int(0.25 * SAMPLE_RATE))
pieces = []

# 3. Execução (Loop Advanced)
print(f"🎬 Gerando áudio para o Drive em: {path_drive}")

for sentence in sentences:
    print(f"🎙️ Processando: {sentence[:50]}...")

    # Parâmetros de controle para evitar alucinações
    semantic_tokens = generate_text_semantic(
        sentence,
        history_prompt=SPEAKER,
        temp=0.6,
        min_eos_p=0.05,
    )

    audio_array = semantic_to_waveform(semantic_tokens, history_prompt=SPEAKER)
    pieces += [audio_array, silence.copy()]

# 4. Concatenar e Salvar
final_array = np.concatenate(pieces)

# Nome do arquivo (você pode mudar aqui)
nome_arquivo = "narracao_guga_final.mp3"
temp_wav = "temp_output.wav"
caminho_final = os.path.join(path_drive, nome_arquivo)

# Exportação para MP3
# Convertemos para 16-bit PCM primeiro
wavfile.write(temp_wav, SAMPLE_RATE, (final_array * 32767).astype(np.int16))
AudioSegment.from_wav(temp_wav).export(caminho_final, format="mp3", bitrate="192k")
os.remove(temp_wav)

print(f"✅ Sucesso! Arquivo salvo em: {caminho_final}")
Audio(caminho_final, autoplay=True)
```
