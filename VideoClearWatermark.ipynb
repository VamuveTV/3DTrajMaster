```python
# 2. Montar o Google Drive
from google.colab import drive
import os
from pathlib import Path

drive.mount('/gdrive')
```

```python
# 3. Configurar os diretórios (Customize aqui se quiser)
# Ele vai criar as pastas automaticamente se elas não existirem
in_path = '/gdrive/MyDrive/VideoClear/Input'
out_path = '/gdrive/MyDrive/VideoClear/Output'

os.makedirs(in_path, exist_ok=True)
os.makedirs(out_path, exist_ok=True)
```

**prepare videos**

```python
# 4. Função para processar os vídeos
def process_videos():
    # Lista todos os arquivos .mp4 na pasta de entrada
    videos = [f for f in Path(in_path).glob("*.mp4")]

    if not videos:
        print(f"Nenhum arquivo .mp4 encontrado em {in_path}")
        return

    print(f"Encontrados {len(videos)} vídeos para processar.")

    for video_file in videos:
        input_file = str(video_file)
        # Define o nome do arquivo de saída mantendo o original + _restored
        output_file = os.path.join(out_path, f"restored_{video_file.name}")

        print(f"--- Processando: {video_file.name} ---")

        # Executa o comando do Sora2WatermarkRemover
        # Ajustei os caminhos para ler do Drive e salvar no Drive
        !python remwm.py "{input_file}" "{output_file}" --max-bbox-percent=15 --force-format=MP4

        print(f"--- Finalizado: {output_file} ---")
```

```python
# 5. Rodar o processo
process_videos()
```
